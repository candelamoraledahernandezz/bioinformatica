```python
# ==========================================================
# CELDA 12: VALIDACIÓN, COMPOSICIÓN, GC Y COMPLEMENTARIA
# ==========================================================

def validar_adn(secuencia):
    # secuencia: texto con la muestra a comprobar.
    # Devuelve True solo si TODAS las letras son A, T, C o G.
    secuencia = secuencia.upper()          # acepta minúsculas
    if len(secuencia) == 0:                # una secuencia vacía no es válida
        return False
    for letra in secuencia:                # letra: cada carácter, uno a uno
        if letra not in NUCLEOTIDOS_VALIDOS:
            return False
    return True

def analizar_composicion(secuencia):
    # Devuelve un diccionario con el número de A, T, C y G.
    secuencia = secuencia.upper()
    return {base: secuencia.count(base) for base in ["A", "T", "C", "G"]}

def calcular_gc(secuencia):
    # Devuelve el porcentaje de G + C sobre la longitud total.
    secuencia = secuencia.upper()
    if len(secuencia) == 0:
        return 0.0
    gc = secuencia.count("G") + secuencia.count("C")   # gc: nº de bases G y C
    return round(gc / len(secuencia) * 100, 2)

def obtener_complementaria(secuencia):
    # Cambia cada base por su pareja usando el diccionario COMPLEMENTARIO.
    secuencia = secuencia.upper()          # acepta minúsculas
    if not validar_adn(secuencia):         # si hay letras raras, evitamos un KeyError
        return ""                          # cadena vacía = no se puede calcular
    return "".join(COMPLEMENTARIO[base] for base in secuencia)

# Prueba rápida con la muestra del Caso 1
print(validar_adn(muestra_caso1))
print(analizar_composicion(muestra_caso1))
print(calcular_gc(muestra_caso1))
print(obtener_complementaria(muestra_caso1))
```

```python
# ==========================================================
# CELDA 13: BÚSQUEDA DE PATRONES Y BIOMARCADORES
# ==========================================================

def buscar_patron(secuencia, patron):
    # secuencia: ADN donde se busca. patron: fragmento que se busca.
    # Devuelve la lista de posiciones (empezando en 0) donde aparece el patrón.
    posiciones = []                                   # posiciones: aciertos encontrados
    for i in range(len(secuencia) - len(patron) + 1): # i: posición de inicio a probar
        if secuencia[i:i + len(patron)] == patron:
            posiciones.append(i)
    return posiciones

def detectar_biomarcador(secuencia, biomarcador):
    # Devuelve True si el biomarcador está dentro de la secuencia.
    return biomarcador in secuencia

# Prueba rápida con las muestras del Caso 2
for letra, seq in muestras_caso2.items():
    print(letra, detectar_biomarcador(seq, biomarcador_caso2), buscar_patron(seq, biomarcador_caso2))
```

```python
# ==========================================================
# CELDA 14: COMPARACIÓN Y MUTACIONES
# ==========================================================
import difflib   # librería estándar de Python; alinea secuencias de distinta longitud

def comparar_secuencias(seq1, seq2):
    # Compara posición por posición.
    # coincidencias: nº de posiciones con la misma letra.
    # longitud: la de la secuencia más larga (si difieren, se penaliza).
    # similitud: (coincidencias / longitud) x 100.
    coincidencias = sum(1 for a, b in zip(seq1, seq2) if a == b)
    longitud = max(len(seq1), len(seq2))
    if longitud == 0:
        return 0, 0.0
    similitud = round(coincidencias / longitud * 100, 2)
    return coincidencias, similitud

def detectar_mutaciones(referencia, muestra):
    # Devuelve una lista de diccionarios, uno por cada alteración:
    # {"tipo", "posicion" (empieza en 1), "referencia", "muestra"}
    mutaciones = []

    # Misma longitud -> solo puede haber sustituciones
    if len(referencia) == len(muestra):
        for i in range(len(referencia)):
            if referencia[i] != muestra[i]:
                mutaciones.append({"tipo": "Sustitución", "posicion": i + 1,
                                   "referencia": referencia[i], "muestra": muestra[i]})
        return mutaciones

    # Distinta longitud -> hay inserciones o deleciones
    comparador = difflib.SequenceMatcher(None, referencia, muestra)
    for etiqueta, i1, i2, j1, j2 in comparador.get_opcodes():
        if etiqueta == "replace":
            mutaciones.append({"tipo": "Sustitución", "posicion": i1 + 1,
                               "referencia": referencia[i1:i2], "muestra": muestra[j1:j2]})
        elif etiqueta == "insert":
            mutaciones.append({"tipo": "Inserción", "posicion": j1 + 1,
                               "referencia": "", "muestra": muestra[j1:j2]})
        elif etiqueta == "delete":
            mutaciones.append({"tipo": "Deleción", "posicion": i1 + 1,
                               "referencia": referencia[i1:i2], "muestra": ""})
    return mutaciones

# Prueba rápida con los tres ejemplos del documento
print(comparar_secuencias(referencia_caso3, muestra_caso3))
for nombre, datos in ejemplos_mutacion.items():
    print(nombre, detectar_mutaciones(datos["referencia"], datos["muestra"]))
```

```python
# ==========================================================
# CELDA 15: PUNTUACIÓN E INFORME FORENSE
# ==========================================================

def calcular_estrellas(intentos, pistas):
    # intentos: errores cometidos en la pregunta. pistas: pistas pedidas.
    # 0 errores -> 3★, 1 -> 2★, 2 -> 1★, 3 -> 0★. Cada pista resta otra estrella.
    if intentos >= MAX_INTENTOS:
        return 0
    return max(0, ESTRELLAS_MAXIMAS - intentos - pistas * PENALIZACION_PISTA)

def mostrar_estrellas(n):
    # n: nº de estrellas. Devuelve ★★★ o ☆ si es 0 (como en el documento).
    return "★" * n if n > 0 else "☆"

def actualizar_informe(pista):
    # pista: texto de la evidencia. Se añade a pistas_desbloqueadas sin duplicarla.
    if pista not in pistas_desbloqueadas:
        pistas_desbloqueadas.append(pista)
        print("📄 Informe forense actualizado:", pista)

def consultar_informe():
    # Muestra todas las pistas conseguidas hasta ahora.
    print("\n--- INFORME FORENSE ---")
    if len(pistas_desbloqueadas) == 0:
        print("(vacío: aún no hay pistas)")
    for pista in pistas_desbloqueadas:
        print(pista)
    print("-----------------------")

# Prueba rápida
print(mostrar_estrellas(calcular_estrellas(0, 0)), mostrar_estrellas(calcular_estrellas(1, 0)),
      mostrar_estrellas(calcular_estrellas(3, 0)))
```

```python
# ==========================================================
# CELDA 16: AUXILIARES DE CONSOLA
# ==========================================================

def linea(caracter="=", n=40):
    # Dibuja una línea decorativa (caracter repetido n veces).
    print(caracter * n)

def titulo(texto):
    # Muestra un título centrado entre dos líneas (estilo sección 19).
    linea()
    print(texto.center(40))
    linea()

def pedir(texto):
    # Pide un dato por teclado y le quita los espacios sobrantes.
    return input(texto).strip()

def si_no(texto):
    # Pregunta de sí/no. Devuelve True si el jugador escribe s, si o sí.
    return pedir(texto + " (s/n): ").lower() in ("s", "si", "sí")

def a_numero(texto):
    # Convierte lo escrito a número. Acepta coma decimal, espacios y el símbolo %.
    # Ejemplos válidos: "50", "50,0", "50 %", "50%".
    # Devuelve None si no es un número, para evitar errores del programa.
    texto = texto.replace("%", "").replace(" ", "").replace(",", ".")   # limpia el texto
    try:
        valor = float(texto)
    except ValueError:
        return None
    if valor != valor or valor in (float("inf"), float("-inf")):        # rechaza "nan" e "inf"
        return None
    return valor
```

```python
# ==========================================================
# CELDA 17: MOTOR DE PREGUNTAS Y FIN DE CASO
# ==========================================================

def resolver_pregunta(enunciado, es_correcta, solucion_texto, pista):
    # enunciado: la pregunta que se muestra.
    # es_correcta: función que recibe la respuesta y devuelve True/False.
    # solucion_texto: lo que se enseña si se agotan los intentos.
    # pista: ayuda opcional (cuesta 1 estrella).
    # Devuelve (acierto, estrellas).
    global intentos, pistas_usadas, respuestas_correctas
    intentos = 0           # se reinicia en cada pregunta
    pistas_usadas = 0
    print("\n" + enunciado)
    while intentos < MAX_INTENTOS:
        respuesta = pedir("Respuesta: ")
        if es_correcta(respuesta):
            print("✔ ¡Análisis correcto!")
            respuestas_correctas += 1
            return True, calcular_estrellas(intentos, pistas_usadas)
        intentos += 1
        print(f"✘ Incorrecto. Errores: {intentos}/{MAX_INTENTOS}")
        if intentos < MAX_INTENTOS and pistas_usadas == 0:
            if si_no("¿Quieres solicitar una pista?"):
                pistas_usadas += 1
                print("💡 PISTA:", pista, "(-1 estrella)")
    print("Sin intentos. Ayuda del laboratorio:", solucion_texto)
    return False, 0

def finalizar_caso(numero, lista_estrellas):
    # numero: caso terminado. lista_estrellas: estrellas de cada pregunta.
    # Guarda la media (redondeada) y desbloquea el siguiente caso.
    global puntuacion
    resultado = int(sum(lista_estrellas) / len(lista_estrellas) + 0.5)
    estrellas[numero] = max(estrellas[numero], resultado)   # conserva el mejor resultado
    if numero not in casos_completados:
        casos_completados.append(numero)
    puntuacion = sum(estrellas.values())
    print(f"\nResultado del caso {numero}: {mostrar_estrellas(resultado)}")
    siguiente = numero + 1
    if siguiente in casos_desbloqueados and not casos_desbloqueados[siguiente]:
        casos_desbloqueados[siguiente] = True
        print(f"🔓 Caso {siguiente} desbloqueado: {nombres_casos[siguiente]}")
```

```python
# ==========================================================
# CELDA 20: CASO 3
# ==========================================================

def caso_3():
    global caso_actual
    caso_actual = 3
    referencia = secuencias_referencia["caso3"]   # referencia: se lee del diccionario de la celda 6
    titulo("CASO 03 - " + nombres_casos[3])
    print(contexto_casos[3])
    print("\nSECUENCIA DE REFERENCIA:", referencia)
    print("MUESTRA FORENSE:        ", muestra_caso3)
    coinc, sim = comparar_secuencias(referencia, muestra_caso3)
    print(f"Coincidencias: {coinc} | Similitud: {sim} %")
    mutaciones = detectar_mutaciones(referencia, muestra_caso3)   # mutaciones: lista calculada por el programa
    resultados = []

    print("\n¿Qué tipo de alteración observas?")
    for numero, nombre in tipos_mutacion.items():
        print(f"[{numero}] {nombre}")
    ok, e = resolver_pregunta(
        "1) Elige una opción (1-4)",
        lambda r: r == str(respuesta_correcta_caso3),
        f"{respuesta_correcta_caso3} - {tipos_mutacion[respuesta_correcta_caso3]}",
        "Las dos secuencias tienen la misma longitud.")
    resultados.append(e)
    if ok:
        actualizar_informe(textos_pistas["mutacion"])

    pos = mutaciones[0]["posicion"]   # pos: posición de la primera diferencia
    ok, e = resolver_pregunta(
        "2) ¿En qué posición está la diferencia? (la primera letra es la 1)",
        lambda r: a_numero(r) == pos,
        str(pos),
        "Compara las dos secuencias letra a letra.")
    resultados.append(e)

    finalizar_caso(3, resultados)
```

```python
# ==========================================================
# CELDA 21: CASO FINAL
# ==========================================================

def cargar_sospechosos():
    # Muestra la base de datos con la ficha, los biomarcadores y el ADN de cada sospechoso.
    print("\nBASE DE DATOS DE SOSPECHOSOS")
    for codigo, adn in sospechosos.items():
        ficha = info_sospechosos[codigo]                       # ficha: nombre e información del caso
        marcadores = biomarcadores_sospechosos[codigo]         # marcadores: biomarcadores registrados en su historial
        texto_marcadores = ", ".join(marcadores) if marcadores else "ninguno"
        print(f"{codigo} | {ficha['nombre']} | {ficha['info']}")
        print(f"      ADN: {adn}")
        # Nota: el biomarcador es un dato del historial del sospechoso (evidencia del Caso 2),
        # no tiene por qué aparecer dentro de esta secuencia de 16 letras.
        print(f"      Biomarcadores registrados: {texto_marcadores}")
    return sospechosos

def acusacion_final():
    # Pide el código del culpable entre los sospechosos NO descartados.
    # Devuelve (acierto, estrellas).
    restantes = [c for c in sospechosos if c not in sospechosos_descartados]   # restantes: aún bajo sospecha
    print("\nSospechosos restantes:", ", ".join(restantes))
    print("Descartados:", ", ".join(sospechosos_descartados) if sospechosos_descartados else "ninguno")
    return resolver_pregunta(
        "Selecciona al culpable (escribe su código, por ejemplo S01):",
        lambda r: r.upper() == culpable,
        culpable,
        "Descarta a quien no coincida al 100 % con la muestra o no tenga el biomarcador.")

def caso_final():
    global caso_actual
    caso_actual = 4
    titulo("CASO FINAL - " + nombres_casos[4])
    print(contexto_casos[4])
    print("\nMUESTRA FORENSE:", muestra_forense)
    cargar_sospechosos()

    print("\nCOMPARACIÓN GENÉTICA")
    similitudes = {}    # similitudes: código -> % de similitud con la muestra forense
    for codigo, adn in sospechosos.items():
        coinc, sim = comparar_secuencias(muestra_forense, adn)
        similitudes[codigo] = sim
        print(f"{codigo}: {coinc}/{len(muestra_forense)} coincidencias -> {sim} %")
    actualizar_informe(textos_pistas["comparacion"])

    resultados = []     # resultados: estrellas de cada pregunta del caso
    a_descartar = {c for c, s in similitudes.items()
               if s < 100 or biomarcadores["caso2"] not in biomarcadores_sospechosos[c]}
    def descarte_correcto(r):
        # Convierte "S02, S03" en un conjunto y lo compara con a_descartar.
        return {x.strip().upper() for x in r.split(",") if x.strip()} == a_descartar

    ok, e = resolver_pregunta(
        "1) ¿Qué sospechosos descartas? (códigos separados por comas)",
        descarte_correcto,
        ", ".join(sorted(a_descartar)),
        "Descarta a quien tenga alguna diferencia con la muestra forense.")
    resultados.append(e)
    if ok:
        for c in sorted(a_descartar):
            if c not in sospechosos_descartados:
                sospechosos_descartados.append(c)
        actualizar_informe(textos_pistas["descarte"])

    consultar_informe()    # el jugador revisa las pruebas antes de acusar

    ok, e = acusacion_final()
    resultados.append(e)
    if ok:
        print(f"\n🎉 CASO RESUELTO: el culpable es {culpable} ({info_sospechosos[culpable]['nombre']})")
    else:
        print(f"\n❌ Acusación fallida. El culpable era {culpable}.")

    finalizar_caso(4, resultados)
```

```python
# ==========================================================
# CELDA 22: MENÚ Y PANEL DE PROGRESO
# ==========================================================

def instrucciones():
    titulo("INSTRUCCIONES")
    print(texto_instrucciones)

def mostrar_panel_progreso():
    titulo("PANEL DE PROGRESO")
    for n in range(1, TOTAL_CASOS + 1):
        icono = "🔓" if casos_desbloqueados[n] else "🔒"
        resultado = mostrar_estrellas(estrellas[n]) if n in casos_completados else "-"
        print(f"{icono} CASO {n}: {nombres_casos[n]:<30} {resultado}")
    print(f"\nTOTAL: {puntuacion}/{ESTRELLAS_TOTALES_POSIBLES} ★")
    print(f"Respuestas correctas: {respuestas_correctas}")

def mostrar_menu():
    # Muestra el menú principal con el subtítulo centrado.
    titulo("DNA DETECTIVE")
    print("LABORATORIO FORENSE".center(40))   # center(40): mismo ancho que las líneas
    print("[1] Nueva investigación (elegir caso)")
    print("[2] Instrucciones")
    print("[3] Panel de progreso")
    print("[4] Informe forense")
    print("[5] Reiniciar partida")
    print("[6] Salir")

def reiniciar_partida():
    # Devuelve todas las variables de estado (Celda 8) a su valor inicial.
    global caso_actual, intentos, pistas_usadas, puntuacion, estrellas
    global respuestas_correctas, pistas_desbloqueadas, casos_desbloqueados
    global sospechosos_descartados, casos_completados
    caso_actual = 1
    intentos = 0
    pistas_usadas = 0
    puntuacion = 0
    estrellas = {1: 0, 2: 0, 3: 0, 4: 0}
    respuestas_correctas = 0
    pistas_desbloqueadas = []
    casos_desbloqueados = {1: True, 2: False, 3: False, 4: False}
    sospechosos_descartados = []
    casos_completados = []
    print("Partida reiniciada.")

# Relaciona el número de caso con su función, para llamarla desde el menú.
casos = {1: caso_1, 2: caso_2, 3: caso_3, 4: caso_final}

def seleccionar_caso():
    # Muestra el archivo de casos y devuelve el elegido (o None si está bloqueado).
    print("\nARCHIVO DE CASOS")
    for n in range(1, TOTAL_CASOS + 1):
        icono = "🔓" if casos_desbloqueados[n] else "🔒"
        print(f"[{n}] {icono} {nombres_casos[n]}")
    opcion = pedir("Caso: ")
    if opcion.isdigit() and int(opcion) in casos and casos_desbloqueados[int(opcion)]:
        return int(opcion)
    print("Caso no disponible.")
    return None
```

```python
# ==========================================================
# CELDA 23: MAIN
# ==========================================================

def main():
    # Bucle principal: muestra el menú hasta que el jugador elija salir.
    # El try/except evita un error feo si se interrumpe el programa (Ctrl+C o botón ■).
    try:
        while True:
            mostrar_menu()
            opcion = pedir("Opción: ")
            if opcion == "1":
                n = seleccionar_caso()          # n: caso elegido (None si está bloqueado)
                if n:
                    casos[n]()
            elif opcion == "2":
                instrucciones()
            elif opcion == "3":
                mostrar_panel_progreso()
            elif opcion == "4":
                consultar_informe()
            elif opcion == "5":
                # Pide confirmación para no borrar el progreso por error.
                if si_no("¿Seguro que quieres borrar todo el progreso?"):
                    reiniciar_partida()
                else:
                    print("Reinicio cancelado.")
            elif opcion == "6":
                print("¡Hasta pronto, detective!")
                break
            else:
                print("Opción no válida.")
    except (KeyboardInterrupt, EOFError):
        print("\nPartida interrumpida. ¡Hasta pronto, detective!")
```
