```python
import random

# 1. Número secreto entre 1 y 100
numero_secreto = random.randint(1, 100)

# 2. Introducción
print("¡Bienvenido al juego de adivinanzas!")
print("Estoy pensando en un número entre 1 y 100.")
print("Intenta adivinarlo.")
print("Si estás a 10 o menos: ¡CALIENTE!")
print("Si estás a más de 10: ¡FRÍO!")

# 3. Lista para almacenar los intentos
intentos = []

# 4. Bucle del juego
while True:
    intento = int(input("Adivina el número: "))

    # Comprobar límites
    if intento < 1 or intento > 100:
        print("FUERA DE LÍMITES")
        continue

    # Guardar el intento
    intentos.append(intento)

    # Si acierta
    if intento == numero_secreto:
        print("¡HAS ACERTADO!")
        print("Te ha llevado", len(intentos), "intentos.")
        break

    # Primer intento
    if len(intentos) == 1:
        if abs(numero_secreto - intento) <= 10:
            print("¡CALIENTE!")
        else:
            print("¡FRÍO!")

    # Intentos siguientes
    else:
        distancia_actual = abs(numero_secreto - intento)
        distancia_anterior = abs(numero_secreto - intentos[-2])

        if distancia_actual < distancia_anterior:
            print("¡MÁS CALIENTE!")
        else:
            print("¡MÁS FRÍO!")
```
